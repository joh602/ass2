```python
from collections import deque

# CREATE GRAPH
def build_graph():
    graph = {}

    n = int(input("Enter the number of nodes in your graph: "))

    for i in range(n):
        data = input(f"Enter node {i+1} and its neighbors: ")

        neighbours = []
        current = ""

        for char in data:
            if char != " ":
                current += char
            else:
                if current:
                    neighbours.append(current)
                    current = ""

        if current:  # Last item
            neighbours.append(current)

        if neighbours:
            node = neighbours[0]
            neigh = neighbours[1:]
            graph[node] = neigh

    print("Graph is :- ", graph)

    return graph


# DFS
def dfs(graph, start, target):
    visited = set()
    stack = [start]
    visited.add(start)
    ans = []

    while stack:
        cn = stack.pop()
        ans.append(cn)

        print("Current traversal path (ans):", ans)

        if cn == target:
            print(f"Target '{target}' found!")
            break

        for nei in reversed(graph.get(cn, [])):
            if nei not in visited:
                visited.add(nei)
                stack.append(nei)

                print("Current Stack :   ", stack)

    print("Final DFS :", ans)


# BFS
def bfs(graph, start, target):
    visited = set()
    queue = deque([start])
    visited.add(start)
    ans = []

    while queue:
        cn = queue.popleft()
        ans.append(cn)

        print("Current traversal path (ans):", ans)

        if cn == target:
            print(f"Target '{target}' found!")
            break

        for nei in graph.get(cn, []):
            if nei not in visited:
                visited.add(nei)
                queue.append(nei)

                print("Current Queue :   ", queue)

    print("Final BFS :", ans)


# MAIN PROGRAM
graph = build_graph()

while True:
    print("\n1. Depth First Search (DFS)")
    print("2. Breadth First Search (BFS)")
    print("3. Exit")

    choice = input("\nEnter your choice (1-3): ")

    if choice == "1":
        start = input("Enter the start node: ")
        target = input("Enter the target node: ")

        dfs(graph, start, target)

    elif choice == "2":
        start = input("Enter the start node: ")
        target = input("Enter the target node: ")

        bfs(graph, start, target)

    elif choice == "3":
        print("Exiting program.")
        break
```
